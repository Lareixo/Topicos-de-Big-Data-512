## Actividad 3 - Estadistica descriptiva

### Proposito
Aplicar medidas estadisticas basicas a una variable poblacional y analizar su distribucion.

### Instrucciones
Selecciona un ano del periodo historico y filtra los datos correspondientes. Sobre la variable de cambio anual de poblacion, calcula la media, mediana, varianza, desviacion estandar, minimo y maximo.

### Analisis
Se usara el ano 2023 para calcular estos estadisticos sobre la columna 'Annual change in population'.

````python
year = 2023
subset = df[df['Year'] == year].copy()
col = 'Annual change in population'

media = subset[col].mean()
mediana = subset[col].median()
varianza = subset[col].var()
desviacion = subset[col].std()
minimo = subset[col].min()
maximo = subset[col].max()

print(f'Year seleccionado: {year}')
print(f'Media: {media:.2f}')
print(f'Mediana: {mediana:.2f}')
print(f'Varianza: {varianza:.2f}')
print(f'Desviacion estandar: {desviacion:.2f}')
print(f'Minimo: {minimo:.2f}')
print(f'Maximo: {maximo:.2f}')

fila_min = subset.loc[subset[col].idxmin()]
fila_max = subset.loc[subset[col].idxmax()]
print(f'Entidad con el valor minimo: {fila_min["Entity"]} ({minimo:.2f})')
print(f'Entidad con el valor maximo: {fila_max["Entity"]} ({maximo:.2f})')

print('\nRESPUESTAS:')
print(f'- Media: {media:.2f}')
print(f'- Mediana: {mediana:.2f}')
print(f'- Desviacion estandar: {desviacion:.2f}')
print(f'- Entidad con valor minimo: {fila_min["Entity"]} ({minimo:.2f})')
print(f'- Entidad con valor maximo: {fila_max["Entity"]} ({maximo:.2f})')
````
